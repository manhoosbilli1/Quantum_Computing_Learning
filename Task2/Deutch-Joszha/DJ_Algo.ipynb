```python
from qiskit import QuantumCircuit 
from qiskit.quantum_info import Statevector
qc = QuantumCircuit(4,4) 
qc.x(3)
display(qc.draw(output='mpl'))

# Truth table. 
# 000, 001, 010, 011, 100, 101, 110, 111  there are 8 possible inputs. ranging from 0-7 in decimal.
# will run the circuit for each input to and measure output as truth table. 

state0 = Statevector.from_label('0000')
state1 = Statevector.from_label('0001')
state2 = Statevector.from_label('0010')
state3 = Statevector.from_label('0011')
state4 = Statevector.from_label('0100')
state5 = Statevector.from_label('0101')
state6 = Statevector.from_label('0110')
state7 = Statevector.from_label('0111')


evolved0 = state0.evolve(qc)
evolved1 = state1.evolve(qc)
evolved2 = state2.evolve(qc)
evolved3 = state3.evolve(qc)
evolved4 = state4.evolve(qc)
evolved5 = state5.evolve(qc)
evolved6 = state6.evolve(qc)
evolved7 = state7.evolve(qc)

display(Math(r'\text{Truth Table}'))
display(evolved0.draw('latex'))
display(evolved1.draw('latex'))
display(evolved2.draw('latex'))
display(evolved3.draw('latex'))
display(evolved4.draw('latex'))
display(evolved5.draw('latex'))
display(evolved6.draw('latex'))
display(evolved7.draw('latex'))


a, state0_out = evolved0.measure()
b, state1_out = evolved1.measure()

print(display(state0_out.draw))

#
```
